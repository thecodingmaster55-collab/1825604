```python
import numpy as np
from scipy.integrate import solve_ivp
from scipy.optimize import root_scalar

def ode (x,u):
    y,yp = u
    return [yp,2*x+y]
x0,y0 = 0,0
xf,yf =1,1

def residual(slope):
    sol=solve_ivp(ode,[x0,xf],[y0,slope],t_eval=[xf])
    return sol.y[0,-1]-yf

result = root_scalar(residual,bracket=[0,3])
correct_slope = result.root

print(f"correct initial slope y'(0)={correct_slope:.6f}")

sol= solve_ivp(ode,[x0,xf],[y0,correct_slope], t_eval=np.linspace(x0,xf,100))

x=sol.t
y=sol.y[0]
print(f"y(1) achieved = { y[-1]:.6f}")
```
