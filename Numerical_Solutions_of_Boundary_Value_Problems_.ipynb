```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(0, 1, 100)

y = (np.exp(x) + np.exp(2-x))/(1 + np.exp(2))

plt.plot(x, y)

plt.xlabel("Distance x")
plt.ylabel("Concentration y(x)")
plt.title("Steady State Concentration Profile")

plt.grid()
plt.show()

print("Concentration:", y)
```
