```python
import numpy as np

x = [1, 2, 3, 4]

N = len(x)

X = []

for k in range(N):
    s = 0
    for n in range(N):
        s += x[n] * np.exp(-2j * np.pi * k * n / N)
    X.append(s)

print("DFT:", np.round(X, 2))
```
