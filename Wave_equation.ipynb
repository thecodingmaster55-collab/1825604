```python
import numpy as np
import matplotlib.pyplot as plt

nx = 41
dx = 2.0 / (nx - 1)
nt = 20
dt = 0.0002
alpha = 0.01
sigma = alpha * dt / dx**2

u = np.ones(nx)
u[int(0.5 / dx) : int(1 / dx + 1)] = 2.0

for n in range(nt):
  un = u.copy()
  for i in range(1, nx - 1):
    u[i] = un[i] + sigma * (un[i + 1] - 2 * un[i] + un[i - 1])

plt.plot(np.linspace(0, 2, nx), u)
plt.title("Heat Equation Solution")
plt.xlabel("x")
plt.ylabel("u")
plt.grid(True)
plt.show()
```
