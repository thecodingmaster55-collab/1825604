```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-5, 5, 100)

y = np.exp(-x**2)

Y = np.fft.fft(y)

freq = np.fft.fftfreq(len(x), x[1]-x[0])

plt.plot(freq, np.abs(Y))

plt.xlabel("Frequency")
plt.ylabel("Amplitude")
plt.title("Fast Fourier Transform")

plt.show()
```
